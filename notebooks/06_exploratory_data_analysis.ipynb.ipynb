```python
#Call libraries
from pyspark.sql import functions as F

#Load Gold tables
game_features_df = spark.table(
    "workspace.sudoku_gold.game_features"
)

solver_targets_df = spark.table(
    "workspace.sudoku_gold.solver_targets"
)

#Combine features and target metrics
eda_df = game_features_df.join(
    solver_targets_df.select(
        "quizzes",
        "naked_single_steps",
        "hidden_single_steps",
        "naked_pair_eliminations",
    ),
    on="quizzes",
    how="inner",
)

print("Exploratory Data Analysis - Dataset Overview")
print("Total records:", eda_df.count())
print("Total columns:", len(eda_df.columns))

display(eda_df.limit(10))
```
